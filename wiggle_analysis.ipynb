```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

file = 'wiggle_data.csv'
data = pd.read_csv(file)
data
```

```python
# plot data against r_AU
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(10, 4))

ax1.plot(data['r_AU'], data['delta_theta_rad'], "o-")
ax1.invert_xaxis()
ax1.set_xlabel('r (AU)')
ax1.set_ylabel('Delta Theta (rad)')

ax2.plot(data['r_AU'], data['delta_phi_rad'], "o-")
ax2.invert_xaxis()
ax2.set_xlabel('r (AU)')
ax2.set_ylabel('Delta Phi (rad)')

ax3.plot(data['r_AU'], data['orb_ang_rad'], "o-")
ax3.invert_xaxis()
ax3.set_xlabel('r (AU)')
ax3.set_ylabel('Orbit Angle (rad)')

plt.tight_layout()
```
