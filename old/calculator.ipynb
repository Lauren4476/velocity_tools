```python
import jax.numpy as jnp
import astropy.units as u
from astropy.constants import G as Gast
au_in_km = 1.4959787e8 #km
```

```python
ecc = 0.998872
mu = 0.001922
theta0 = 2.2689

theta0_deg = jnp.degrees(theta0)

orb_ang_arg = (1 - mu * jnp.sin(theta0)**2) / ecc
print("orb_ang_arg={0}".format(orb_ang_arg))
orb_ang = jnp.arccos((1 - mu * jnp.sin(theta0)**2) / ecc) #<- this is in radians

print("orb_ang = {0}".format(orb_ang))
```

```python
G = 6.67430e-11 * (1e-3)**2 * (1.988416e30) / (1.4959787e11) # in au (km/s)^2 Msol^-1
print(G)
G_astropy = Gast.to(u.au * (u.km / u.s)**2 / u.M_sun).value
print("G_astropy={0}".format(G_astropy))
```

```python
r0 = 1000 # au
omega0 = 7e-13 # /s
mass = 2.1 # solar masses

def r_cent(mass, omega=1e-14, r0=1e4):
    """
    Centrifugal radius or disk radius in the Ulrich (1976)'s model.
    r_u in Mendoza's nomenclature.

    :param mass: Central mass for the protostar, Msun
    :param omega: Angular speed at the r0 radius, 1/s
    :param r0: Initial radius of the streamline, au
    :return: r_cent, au
    """
    r_cent = (r0 ** 4 * omega ** 2 / (G * mass)) # in au^3 km^-2
    r_cent_au = r_cent * (au_in_km**2) # in au
    print("rc={0} au".format(r_cent_au))
    return r_cent_au

r_cent = r_cent(mass=mass, omega=omega0, r0=r0)
r_cent
```

```python
r0 = 1000*u.au # au
omega0 = 7e-13 / u.s
mass = 2.1 * u.M_sun

r_cent = (r0**4 * omega0**2 / (Gast * mass)).to(u.au)
print("r_cent astropy = {0}".format(r_cent))
```
